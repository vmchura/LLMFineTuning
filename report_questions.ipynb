```python
# Install transformers from source - only needed for versions <= v4.34
# pip install git+https://github.com/huggingface/transformers.git
# pip install accelerate

import torch
from transformers import pipeline

pipe = pipeline("text-generation", model="Qwen/Qwen2-0.5B-Instruct", torch_dtype=torch.bfloat16, device_map="auto")

# We use the tokenizer's chat template to format each message - see https://huggingface.co/docs/transformers/main/en/chat_templating
messages = [
    {"role": "user", "content": "How many waste collection routes are established?"},
]
prompt = pipe.tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
outputs = pipe(prompt, max_new_tokens=256, do_sample=True, temperature=0.7, top_k=50, top_p=0.95)
print(outputs[0]["generated_text"])
# <|system|>
# You are a friendly chatbot who always responds in the style of a pirate.</s>
# <|user|>
# How many helicopters can a human eat in one sitting?</s>
# <|assistant|>
# ...
```

```python

import requests
import json

def query_ollama(prompt, model_name="mymodel"):
    """Query the Ollama API with a prompt."""
    url = "http://localhost:11434/api/generate"
    data = {
        "model": model_name,
        "prompt": prompt,
        "stream": False
    }

    response = requests.post(url, json=data)
    return response.json()["response"]

# Format the same message as in your original code
messages = [
    {"role": "user", "content": "How many waste collection routes are established?"},
]

# Simple formatting for Ollama
system_prompt = "You are a helpful assistant."
formatted_prompt = f"{system_prompt}\n\nUser: {messages[0]['content']}\nAssistant:"

# Query the model
result = query_ollama(formatted_prompt)
print(result)
```
