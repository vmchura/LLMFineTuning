```python
import requests
import pandas as pd
```

```python
class ExecutorPrompt:
    def __init__(self, model):
        self.model = model
        self.url = "http://localhost:11434/api/generate"

    def get_response(self, prompt):
        system_prompt = "You are a helpful assistant."
        formatted_prompt = f"{system_prompt}\n\nUser: {prompt}\nAssistant:"
        data = {
            "model": self.model,
            "prompt": formatted_prompt,
            "stream": False
        }

        response = requests.post(self.url, json=data)
        return response.json()["response"]
```

```python
mymodel_prompt = ExecutorPrompt('mymodel')
qwen_prompt = ExecutorPrompt('qwen2.5:0.5b')
```

```python
def print_responses(model, df, file_output):

    with open(file_output, "w+") as f:
        f.write("\\begin{itemize}\n")

        for row in df.itertuples():
            f.write(f"\\item \\textbf{{{row[1]}}}\n")
            if model is not None:
                response = model.get_response(row[1])
            else:
                response = row[2]
            f.write(response)
            f.write("\n")

        f.write("\\end{itemize}")
```

```python
df = pd.read_csv("question_answer.csv", header=0)
```

```python
print_responses(mymodel_prompt, df, 'report/responses_questions/model_B.tex')
```

```python
print_responses(qwen_prompt, df, 'report/responses_questions/model_A.tex')
```

```python
print_responses(None, df, 'report/responses_questions/actual_responses.tex')
```
